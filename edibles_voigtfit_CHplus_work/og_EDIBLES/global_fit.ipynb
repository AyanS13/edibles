```python
from astrovoigtfit import dual_spectrum_voigt_fit
from astrovoigtfit import fit_continuum
from astrovoigtfit import astro_voigt_fit
import numpy as np
import matplotlib.pyplot as plt

from edibles.utils.edibles_oracle import EdiblesOracle
from edibles.utils.edibles_spectrum import EdiblesSpectrum
```

```python
from edibles.utils.final_fitter import *
import matplotlib.gridspec as gridspec
pythia = EdiblesOracle()
List = pythia.getFilteredObsList(object=["HD 147889"], MergedOnly=False, Wave=4044)
test = List.values.tolist()
print(test)
filename = test[2]
print(filename)


# wrange = [6707.1,6708.200000000001]
wrange = [4043.7, 4044.4]
sp = EdiblesSpectrum(filename)
sp.getSpectrum(xmin=wrange[0], xmax=wrange[1])
wave = sp.bary_wave
flux = sp.bary_flux

# Filter the spectrum within the specified wavelength ranges
idx = np.where((wave > wrange[0]) & (wave < wrange[1]))
wave_1 = wave[idx]
flux = flux[idx]

# Normalize the flux
flux_1 = flux / np.median(flux)

restwave_1 = [4044.142]
f_1 = [0.00568]
gamma_1 = [3.79E+07]
N_1=[4.8e+12]
b_1=[2]
v_rad_1 = [-8.2]
map_1 = [0]

plt.plot(wave_1,flux_1)
```

```python
pythia = EdiblesOracle()
List = pythia.getFilteredObsList(object=["HD 147889"], MergedOnly=False, Wave=7699)
test = List.values.tolist()
print(test)
filename = test[0]
print(filename)


#++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
# Define the wavelength range and fetch the spectrum
# wrange = [4035, 4050]
wrange = [7698, 7699.5]
sp = EdiblesSpectrum(filename)
sp.getSpectrum(xmin=wrange[0], xmax=wrange[1])
wave_2= sp.bary_wave
flux_2= sp.bary_flux

# Filter the spectrum within the specified wavelength range
idx = np.where((wave_2 > wrange[0]) & (wave_2< wrange[1]))
wave_2 = wave_2[idx]
flux_2 = flux_2[idx]

# Normalize the flux
flux_2 = flux_2 / np.median(flux_2)


restwave_2 = [7698.965]
f_2 = [0.3331]
gamma_2 = [3.79E+07]
N_2=[4.8e+10]
b_2=[2]
v_rad_2 = [-8.2]
map_2 = [0]

plt.plot(wave_2,flux_2)
```

```python
wavegrid1 = wave_1
wavegrid2 = wave_2
ydata1 = flux_1
ydata2 = flux_2

species_params1 = {
    0: {   # species 0
        'lambda': restwave_1,
        'f': f_1,
        'gamma': gamma_1,
        'b': b_1,
        'N':[4.8881e+12],
        'v_rad': v_rad_1
    }
}

species_params2 = {
    0: {   # species 0
        'lambda': restwave_2,
        'f': f_2,
        'gamma': gamma_2,
        'b': b_2,
        'N': [4.8881e+12],
        'v_rad': v_rad_2
    }
}

result = dual_spectrum_voigt_fit(
    wavegrid1, ydata1, species_params1,
    wavegrid2, ydata2, species_params2
)
```

```python
# Print a detailed fit report
from lmfit import fit_report
print(fit_report(result))
```

```python
for name, param in result.params.items():
    print(f"{name:10s} = {param.value:.4e} ± {param.stderr if param.stderr else 'NA'}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

# Split fit into spectrum 1 and 2
yfit1 = result.best_fit[:len(ydata1)]
yfit2 = result.best_fit[len(ydata1):]

# -------------------------------
# Plot Spectrum 1
# -------------------------------
fig = plt.figure(figsize=(6, 4), dpi=100)
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 1.2])

ax1 = plt.subplot(gs[0])
ax2 = plt.subplot(gs[1])

# Top panel: data + fit
ax1.grid(alpha=0.5)
ax1.plot(wavegrid1, ydata1, c='gray', label='Data')
ax1.plot(wavegrid1, yfit1, color='purple', linestyle='--', label='Fitted Curve')
ax1.set_ylabel("Normalised Flux", fontweight='bold')
ax1.legend()
ax1.set_xlabel("Wavelength ($\\AA$)", fontweight='bold')
ax1.set_xlim(min(wavegrid1), max(wavegrid1))  # adjust manually if needed

# Residuals
residuals1 = ydata1 - yfit1
ax2.scatter(wavegrid1, residuals1, color='green', s=10)
ax2.axhline(0, color='blue', linestyle='--')
ax2.set_xlim(min(wavegrid1), max(wavegrid1))
ax2.set_ylim(1.1*residuals1.min(), 1.1*residuals1.max())
ax2.set_xlabel("Wavelength ($\\AA$)", fontweight='bold')
ax2.set_ylabel("Residuals", fontweight='bold')

plt.tight_layout()
plt.savefig("spectrum1_fit.png", dpi=600)
plt.show()


# -------------------------------
# Plot Spectrum 2
# -------------------------------
fig = plt.figure(figsize=(6, 4), dpi=100)
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 1.2])

ax1 = plt.subplot(gs[0])
ax2 = plt.subplot(gs[1])

# Top panel: data + fit
ax1.grid(alpha=0.5)
ax1.plot(wavegrid2, ydata2, c='gray', label='Data')
ax1.plot(wavegrid2, yfit2, color='purple', linestyle='--', label='Fitted Curve')
ax1.set_ylabel("Normalised Flux", fontweight='bold')
ax1.legend()
ax1.set_xlabel("Wavelength ($\\AA$)", fontweight='bold')
ax1.set_xlim(min(wavegrid2), max(wavegrid2))

# Residuals
residuals2 = ydata2 - yfit2
ax2.scatter(wavegrid2, residuals2, color='green', s=10)
ax2.axhline(0, color='blue', linestyle='--')
ax2.set_xlim(min(wavegrid2), max(wavegrid2))
ax2.set_ylim(1.1*residuals2.min(), 1.1*residuals2.max())
ax2.set_xlabel("Wavelength ($\\AA$)", fontweight='bold')
ax2.set_ylabel("Residuals", fontweight='bold')

plt.tight_layout()
plt.savefig("spectrum2_fit.png", dpi=600)
plt.show()
```
