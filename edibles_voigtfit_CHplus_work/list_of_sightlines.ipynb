```python
import os
import numpy as np

def get_sightlines(directory_path):
    # Check if the directory exists
    if not os.path.exists(directory_path):
        print(f"Directory '{directory_path}' does not exist.")
        return []
    
    # Initialize empty list to store folder names
    sightlines = []
    
    # Iterate through items in the directory
    for item in os.listdir(directory_path):
        # Check if item is a directory and starts with 'HDX'
        if os.path.isdir(os.path.join(directory_path, item)) and item.startswith('HD'):
            sightlines.append(item)
    
    return sightlines

# Example usage
directory_path = "DR4"  # Replace with your folder path
folder_list = get_sightlines(directory_path)
print("All EDIBLES sightlines in DR4:", folder_list)
print("Total number of sightlines:", len(folder_list))
```

```python
import matplotlib.pyplot as plt

def column_plot(x_data, y_data, xlabel='X', ylabel='Y', title=None):
    """
    Creates a simple column/bar plot with x and y data.
    
    Parameters:
        x_data: array-like - data for x-axis
        y_data: array-like - data for y-axis
        xlabel: str - x-axis label
        ylabel: str - y-axis label
        title: str - plot title (optional)
    """
    plt.figure(figsize=(8, 6))
    plt.bar(x_data, y_data, color='skyblue', edgecolor='black')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    if title:
        plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.show()

# Example usage with some sample data:
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([21, 18, 39, 19, 12, 6])
column_plot(x, y, xlabel='Number of Clouds', ylabel='Number of Sightlines')
```

```python
'''
Tallying the sightlnes which have only one O15 data file
'''

import numpy as np
import matplotlib.pyplot as plt
from edibles.utils.edibles_oracle import EdiblesOracle
from IPython.display import display, Math
from scipy.interpolate import CubicSpline

single_file_sightlines = []

for s in folder_list:
    b = s.split('HD')[1]
    sightline = 'HD' + ' ' + b
    pythia = EdiblesOracle()
    List_1 = pythia.getFilteredObsList(object=[sightline], MergedOnly=False, Wave=4232.0)
    test_1 = List_1.values.tolist()
    List_2 = pythia.getFilteredObsList(object=[sightline], MergedOnly=True, Wave=4232.0)
    test_2 = List_2.values.tolist()
    test = set(test_1) - set(test_2)
    test = list(test)

    if s not in bad_sightlines and s not in no_signature_sightlines:
        if len(test) == 2:
            single_file_sightlines.append(s)
```

```python

print("Sightlines with only one O15 data file:", single_file_sightlines)
print("Total number of such sightlines:", len(single_file_sightlines))

multi_line_sightlines = set(folder_list) - set(single_file_sightlines) - bad_sightlines - no_signature_sightlines
print("Sightlines with multiple O15 data files:", multi_line_sightlines)
print("Total number of such sightlines:", len(multi_line_sightlines))

print('unusable sightlines (bad + no CH+ signature):', len(bad_sightlines) + len(no_signature_sightlines))

print('Total categorized sightlines including unusable:', len(single_file_sightlines) + len(multi_line_sightlines) + len(bad_sightlines) + len(no_signature_sightlines))
```
