```python
import os
import numpy as np

def get_sightlines(directory_path):
    # Check if the directory exists
    if not os.path.exists(directory_path):
        print(f"Directory '{directory_path}' does not exist.")
        return []
    
    # Initialize empty list to store folder names
    sightlines = []
    
    # Iterate through items in the directory
    for item in os.listdir(directory_path):
        # Check if item is a directory and starts with 'HDX'
        if os.path.isdir(os.path.join(directory_path, item)) and item.startswith('HD'):
            sightlines.append(item)
    
    return sightlines

# Example usage
directory_path = "DR4"  # Replace with your folder path
folder_list = get_sightlines(directory_path)
print("All EDIBLES sightlines in DR4:", folder_list)
print("Total number of sightlines:", len(folder_list))
```
